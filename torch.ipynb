```python
import torch
a=torch.tensor([[1,2],[3,4],[5,6]])
b=torch.tensor([[7,8],[9,10],[11,12]])
print(torch.cat((a,b),dim=0))
print(torch.cat((a,b),dim=1))
print(torch.cat((a,b),dim=-2))
print(torch.cat((a,b),dim=-1))
a=torch.tensor([[3,5,4],[90,65,56],[23,45,36]])
b=torch.tensor([[79,83],[99,103],[110,86]])
print(torch.cat((a,b),dim=1))
a=torch.tensor([[1,2],[3,4],[5,6]])
b=torch.tensor([[7,8],[9,10]])
print(torch.cat((a,b),dim=0))
```

```python
import torch
a=torch.tensor([1,10,12,13,15,18])
print(torch.split(a,5))
print(torch.split(a,2))
print(torch.split(a,3))
b=torch.arange(20)
print(torch.split(b,[3,7,5,4,1,0]))
print(torch.chunk(b,3))
print(torch.chunk(b,4))
print(torch.chunk(b,5))
a=torch.zeros(4,5,6)
print(a)
print(a.size(1))
a=torch.ones(2,3,6)
print(a)
print(a.size(1))
a=[1,2,3,4,5]
print(a[2])
```

```python
import torch
a=torch.tensor([[5,6,7],[8,9,10]],dtype=float)
print(a.sum())
print(a.sum(dim=0))
print(a.sum(dim=1))#column
print(a.mean())
print(a.mean(dim=0))
print(a.mean(dim=1))
print(a.max())
print(a.max(dim=0))
print(a.max(dim=1))
print(a.argmax())
print(a.argmin())
```

```python
import torch
a=torch.tensor([[10,4,3],[8,3,2]])
b=torch.tensor([[3,5],[4,8],[1,3]])
print(torch.matmul(a,b))#matrix multiplication
print(a@b)#multiplication
```
