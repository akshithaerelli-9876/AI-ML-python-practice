```python
import tensorflow as tf
a=tf.constant([[3,4,5],[6,5,7],[8,7,6]])
print(a)
print(a[1,1])
print(a[0])
b=tf.transpose(a)
print(b)
print(a[2])
```

```python
import tensorflow as tf
a=tf.constant([[3,4,5,7],[6,5,8,7],[8,9,7,6]])
print(a)
print(a[2:3])
print(a[2:3,0:4])
print(a[1:3,2:4])
print(a[0:2,1:3])
print(a[-1,-3])
print(a[-1:-2:-1,-3:-4:-1])
```

```python
import tensorflow as tf
a=tf.constant([[3,4,5,7],[8,9,10,11],[12,13,14,15]])
b=tf.reverse(a,axis=[0])
print(b)
b=tf.reverse(a,axis=[1])
print(b)
b=tf.reverse(a,axis=[1,0])
print(b)
d=tf.reshape(a,(4,-1))
print(d)
d=tf.reshape(a,(-1))
print(d)
```

```python
import tensorflow as tf
a=tf.constant([2,3,4])
b=tf.constant([6,7,8])
print(tf.add(a,b))
print(tf.subtract(a,b))
print(tf.multiply(a,b))
print(tf.divide(a,b))
```
