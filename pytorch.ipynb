```python
import torch
a=torch.tensor([[2,8,10],[7,8,6],[10,11,12]])
b=torch.tensor([[1,3,9],[6,7,8],[8,9,10]])
print(a.sum())
print(torch.sum(a))
print(torch.subtract(a,b))
print(torch.argmax(a))#index position
print(torch.argmin(a))
print(torch.multiply(a,b))
print(torch.divide(a,b))
```
