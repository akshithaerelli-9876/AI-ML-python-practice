```python
import matplotlib.pyplot as plt
RK=[8,7,13,24,32,50]
Tier1=["IIT DHONBAD(JHARKAND)","IIT ROORKEE(UTTARKHAND)","IIT MADRAS(CHENNAI)","IIT KHARAGPUR(WEST BENGAL)","IIT GUWAHAT(ASSAM)","IIT HYDERABAD(TELANGANA)"]
Tier2=["IIIT DM(KURNOOL)","IIIT KALYANI(WEST BENGAL)","IIIT LUCKNOW","IIIT SRICITY(A.P)","IIIT HYDERABAD(T.G)","IIIT ONGOLE(A.P)"]
#plt.plot(RK,Tier1)
#plt.plot(RK,Tier2)
plt.plot(Tier1,RK,marker="o",
        label="IIT College")
plt.plot(Tier2,RK,linestyle="--",color="Purple",
        label="IIIT College")
plt.legend()
plt.xticks(rotation=90)
plt.title("NTRF OF IIT AND IIIT COLLEGES")
plt.ylabel("Ranking positions")
plt.xlabel("Colleges Rankings")
plt.show()
```

```python
#types of charts
#calculate weekly temperature 
import matplotlib.pyplot as plt
days=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
temp=[30,39,32,36,34,38,33]
plt.bar(days,temp)
plt.title("WEEKLY TEMPERATURE")
plt.xlabel("Days")
plt.xticks(rotation=90)
plt.ylabel("Temperature")
plt.show()
```

```python
import matplotlib.pyplot as plt
items=["crab","prawns","fish","octopus","starfish","chicken","mutton"]
price=[350,200,300,800,700,320,1000]
plt.pie(price,labels=items,autopct="%1.1f%%")
plt.title('purchase of non-veg items')
plt.show()
```
